## Project Description
This project is a beginner-level AI assistant that answers questions from a handbook using a simple Retrieval Augmented Generation (RAG) concept.

The system:

- Stores text data
- Retrieves relevant information
- Answers user queries

## Features
- Beginner-friendly
- Simple retrieval system
- Uses handbook-based answers
- Includes prompt engineering

```python
# Creating handbook.txt file

handbook_text = """
Artificial Intelligence (AI) is the simulation of human intelligence by machines.

Machine Learning is a branch of AI that allows systems to learn from data.

Python is a popular programming language used in AI and Data Analytics.

RAG stands for Retrieval Augmented Generation.

Data Analytics helps in analyzing and understanding data.

FAISS is used for vector similarity search.

Prompt Engineering is the process of designing effective prompts for AI models.

Large Language Models can generate human-like text responses.

Students use AI assistants to quickly retrieve information from documents.

This handbook contains information related to AI, Python, Machine Learning, and RAG systems.
"""

with open("handbook.txt", "w") as file:
    file.write(handbook_text)

print("handbook.txt file created successfully!")
```

```python
SYSTEM_PROMPT = """
You are an academic handbook assistant.

Rules:
1. Answer only from handbook data.
2. Give short and simple answers.
3. If answer is not found, say:
   'The handbook does not contain this information.'
"""

print(SYSTEM_PROMPT)
```

```python
# Load handbook data

with open("handbook.txt", "r") as file:
    data = file.readlines()

print("Handbook Loaded Successfully!")
```

```python
print("===================================")
print(" AI Handbook Assistant using RAG ")
print("===================================")

while True:

    question = input("\nAsk your question (type exit to quit): ")

    if question.lower() == "exit":
        print("\nThank You!")
        break

    found = False

    print("\nSearching handbook...\n")

    for line in data:

        if question.lower() in line.lower():

            print("Answer:")
            print(line.strip())

            found = True
            break

    if found == False:
        print("The handbook does not contain this information.")
```
